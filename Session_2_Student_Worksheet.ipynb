```python
csv_url = '/content/FastFoodNutritionMenuV2.csv'
csv_url
```

```python
import pandas as pd
fast_food = pd.read_csv(csv_url)
fast_food
```

```python
fast_food.dtypes
```

```python
numeric_columns = ['Calories', 'Calories from\nFat', 'Total Fat\n(g)',
                   'Saturated Fat\n(g)', 'Trans Fat\n(g)', 'Cholesterol\n(mg)',
                   'Sodium \n(mg)', 'Carbs\n(g)', 'Fiber\n(g)', 'Sugars\n(g)',
                   'Protein\n(g)', 'Weight Watchers\nPnts']

for col in numeric_columns:
    fast_food[col] = pd.to_numeric(fast_food[col], errors='coerce')
```

```python
fast_food.dtypes
```

```python
top_10_fat = fast_food.sort_values('Calories from\nFat', ascending=False).head(10)
top_10_fat
```

```python
top_10_fat['Label'] = top_10_fat['Company'] + ' - ' + top_10_fat['Item']
```

```python
plot = top_10_fat.plot(kind='barh', x='Label', y='Calories from\nFat', figsize=(12,6),
                      title='Fattiest Fast Food Items (Top 10)')
```
